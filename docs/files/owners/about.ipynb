# Wiki3: AI Owners Club — Inaugural Workshop

## *Wednesday June 24th, 2026 @ 6pm PT*

### *Free · Hybrid · In-Person at Hacker Dojo + Online*

# [OwnersClub.ai](https://ownersclub.ai)



***

This is a hands-on workshop where you'll learn to run AI models on your own computer — no subscription required, your data stays on your machine. We'll focus on practical tools you can install and use today, with live demos, guided installations, and time to experiment with your own projects.

The **[AI Owners Club](https://ownersclub.ai) (https://ownersclub.ai)** is for people who want to own their AI tools. This first session covers local-first AI basics and gets you up and running with [Jan.ai](https://jan.ai/), a privacy-first ChatGPT alternative, plus an introduction to building custom AI workflows with your own documents and data.

***

## Prelinary Inugural Workshop Agenda (2.5 hours)

- **Quick [Jan.ai](https://jan.ai/) (https://jan.ai) examples** — See what's possible with local AI right out of the box
- **What, Why, How: Local-First AI** — Why running models locally matters for privacy, cost, and control
- **Open Source & Freedom** — Why OSS and "free as in freedom" are essential to owning your AI
- **[Jan.ai](https://jan.ai/) Demo** — Live walkthrough of features, models, and real-world usage
- **Installation Walkthroughs**
  - Mac setup
  - Windows setup
- **Using Jan Skills** — Extend Jan with web search and research capabilities
- **MCP Servers and Tools** — Introduction to the Model Context Protocol for connecting AI to your apps and data
- **Q&A** — Your questions answered
- **Practice/Problem-Solving Time** — Bring a task or use case you want to tackle; we'll work through it together
- **Member Demos** — Share what you built or learned during the session (optional)
- **Review & Feedback** — What should we cover next? What could be improved?
