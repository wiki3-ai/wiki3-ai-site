# Understanding the Technology
### What is an LLM?

* LLMs are pure functions in software systems which are otherwise pretty much just ordinary applications.
* They solve the incredibly difficult problem of understanding and using human language.
* They are computationally expensive.
* There are wide variations in performance with big factors including size, training data & method, and inference algorithm.

# Solving the RAG Problem
### Adding Document Intelligence to Jan.ai

* **The Problem:** Jan.ai does not support built-in RAG, meaning you can't drop a folder of PDFs into it to ask questions directly.
* **The Solution:** Layer Jan.ai as a clean, privacy-first UI over Haystack and Open WebUI.
* Haystack indexes documents locally, and Open WebUI provides RAG via an OpenAI-compatible API to connect with Jan.ai, keeping all data on your machine.
