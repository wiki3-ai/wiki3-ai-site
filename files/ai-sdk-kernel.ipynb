# AI SDK Chat Kernel

*If you are in the wiki view (static HTML rendering), click "edit" in the top right to open this [JupyterLite](https://jupyterlite.readthedocs.io/) notebook in your browser.*

This notebook demonstrates using the AI SDK Chat Kernel.  It uses the Vercel AI SDK (https://ai-sdk.dev/) to provision LLM models that operate in your choice of web browser, local hosting, and cloud services.

Source for this JupyterLite kernel, "AI SDK Chat" (check that it is selected in the kernel picker above), is located at https://github.com/wiki3-ai/ai-sdk-chat-kernel and is [Apache v2 licensed](https://github.com/wiki3-ai/ai-sdk-chat-kernel?tab=License-1-ov-file).

Three local LLM providers are configured by default, other providers need configuration using the `%chat` magic.
Initialization will be attempted in this order:

1) [Chrome/Edge AI Built-in AI](https://developer.chrome.com/docs/ai/prompt-api) (`%chat provider built-in-ai/core`) requires that the, currently experimental, built-in AI feature is enabled.
   * Enable experimental flag: [chrome://flags/#prompt-api-for-gemini-nano](chrome://flags/#prompt-api-for-gemini-nano) and relaunch.
3) MLC WebLLM ([webllm.mlc.ai](https://webllm.mlc.ai/)) Provider (`%chat provider built-in-ai/webllm`) requires that [WebGPU](https://en.wikipedia.org/wiki/WebGPU) be available in the browser.
   * Default model is `%chat model SmolLM2-135M-Instruct-q0f16-MLC`, details at (https://huggingface.co/mlc-ai/SmolLM2-135M-Instruct-q0f16-MLC).
4) [HuggingFace Transformers.js](https://huggingface.co/docs/transformers.js) (`%chat provider built-in-ai/transformers`) requires that enough memory is available in the browser.
   * Default model is `%chat model HuggingFaceTB/SmolLM2-360M-Instruct`, details at (https://huggingface.co/HuggingFaceTB/SmolLM2-360M-Instruct).
   * Unfortunately there are many common mobile devices and browsers that Transformers.js doesn't work in yet so "Aw snap!" is a common occurence.
  
Those AI SDK Providers are all by [Jakob Hoeg Mørk](https://www.linkedin.com/in/jakob-hoeg-moerk/).  Source at https://github.com/jakobhoeg/built-in-ai and are documented here: https://ai-sdk.dev/providers/community-providers/built-in-ai

## Chrome/Edge AI Built-in AI Provider

Chrome/Edge Built-In AI with the Prompt API (Gemini Nano) [https://developer.chrome.com/docs/ai/prompt-api](https://developer.chrome.com/docs/ai/prompt-api).

This uses the [@built-in-ai/core]() NPM package.

**Requirements:**
- Chrome browser version 138+ 
- Enable experimental flags:
  - [chrome://flags/#prompt-api-for-gemini-nano](chrome://flags/#prompt-api-for-gemini-nano)
  - [chrome://flags/#optimization-guide-on-device-model](chrome://flags/#optimization-guide-on-device-model)
- Restart browser
  - [chrome://restart](chrome://restart)
- First use requires downloading the Gemini Nano model which may take a while because the model is around 2GB.
