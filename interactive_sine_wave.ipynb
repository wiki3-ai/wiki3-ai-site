# Interactive Sine Wave Explorer

This notebook creates an **interactive plot** of a sine wave function using:
- `matplotlib` for plotting
- `ipywidgets` for interactive sliders
- `numpy` for numerical computation

Use the sliders below the plot to adjust **amplitude**, **frequency**, and **phase** in real time!

```python
# First, install ipywidgets if you haven't already
# Uncomment the line below if needed:
# !pip install ipywidgets

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, fixed
import ipywidgets as widgets

# Enable interactive matplotlib output
%matplotlib inline

print("Libraries loaded successfully!")
```

```python
# Define the function to plot
def plot_sine_wave(amplitude=1.0, frequency=1.0, phase=0.0, x_max=10.0):
    """
    Plot a sine wave: y = amplitude * sin(frequency * x + phase)
    
    Parameters:
    - amplitude: Height of the wave
    - frequency: Number of cycles per unit x
    - phase: Horizontal shift (in radians)
    - x_max: Upper limit of the x-axis
    """
    # Generate x values
    x = np.linspace(0, x_max, 500)
    
    # Compute y values
    y = amplitude * np.sin(frequency * x + phase)
    
    # Also compute the envelope
    y_upper = amplitude * np.ones_like(x)
    y_lower = -amplitude * np.ones_like(x)
    
    # Create the plot
    fig, ax = plt.subplots(figsize=(10, 5))
    
    # Plot the sine wave
    ax.plot(x, y, 'b-', linewidth=2.5, label=f'{amplitude:.2f} * sin({frequency:.2f}x + {phase:.2f})')
    
    # Plot the envelope (dashed lines)
    ax.plot(x, y_upper, 'r--', linewidth=0.8, alpha=0.5)
    ax.plot(x, y_lower, 'r--', linewidth=0.8, alpha=0.5)
    
    # Fill between for visual appeal
    ax.fill_between(x, y, alpha=0.15, color='steelblue')
    
    # Labels and styling
    ax.set_xlabel('x', fontsize=12)
    ax.set_ylabel('y', fontsize=12)
    ax.set_title('Interactive Sine Wave', fontsize=14, fontweight='bold')
    ax.set_xlim(0, x_max)
    ax.set_ylim(-amplitude * 1.5 if amplitude > 0 else -2, amplitude * 1.5 if amplitude > 0 else 2)
    ax.grid(True, alpha=0.3)
    ax.legend(loc='upper right', fontsize=10)
    
    plt.tight_layout()
    plt.show()
```

```python
# Create interactive widgets!
# The @interact decorator automatically generates sliders

_ = interact(
    plot_sine_wave,
    amplitude=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description='Amplitude'),
    frequency=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description='Frequency'),
    phase=FloatSlider(min=0.0, max=2*np.pi, step=0.1, value=0.0, description='Phase (rad)'),
    x_max=fixed(10.0)  # fixed parameter, not adjustable via slider
)

print("\n✅ Use the sliders above to interact with the plot!")
```

## What's happening?

| Parameter | Effect |
|-----------|--------|
| **Amplitude** | Controls the height (peak-to-peak) of the wave |
| **Frequency** | Controls how many oscillations fit in the window |
| **Phase** | Shifts the wave left/right along the x-axis |

The red dashed lines show the **envelope** (the maximum extent of the wave).
